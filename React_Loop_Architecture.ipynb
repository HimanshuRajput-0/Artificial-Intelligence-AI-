# Intro:

The ReAct(Reason and Act) pattern is one of the most important architectural pattern in modern AI agent design.

The React pattern was introduced in the 2023 research paper "React: Synergizing Reasoning and Acting in Language Models"

The core insight is simple: LLMs perform better when they alternate between reasoning(thinking about what to do) and acting(using tools to gather information).

Without this alternation, agents tend to either hallucinate answers(reasoning without acting) or blindly execute tools without understanding the results(acting without reasoning).

Read a result, think about what you learned, decide what to search next, and repeat until you have a complete picture. That is exactly how an AI agent works when it uses the Reactloop Pattern.

Here is how the React loop works at a high level:

1. Receive a research question from the user.

2. The LLM(the brain) reasons about the questions and decides which took to call.

3. The tool executor(the hands) runs the selected tool and returns observations.

4. A router checks whether the LLM wants to call more tools or is done.

5. If more tools are needed, loop back to step 2.

6. If done, compress the research into a final answer.
